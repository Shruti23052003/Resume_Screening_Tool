```python
from PyPDF2 import PdfReader
from docx import Document

def extract_resume_text(file_path):

    extension = os.path.splitext(file_path)[1].lower()

    if extension == ".pdf":

        reader = PdfReader(file_path)

        text = ""

        for page in reader.pages:
            page_text = page.extract_text()

            if page_text:
                text += page_text + "\n"

        return text

    elif extension == ".docx":

        document = Document(file_path)

        text = "\n".join(
            paragraph.text
            for paragraph in document.paragraphs
        )

        return text

    else:
        raise ValueError(
            "Only PDF and DOCX files are supported."
        )
```

```python
def calculate_skill_match(candidate_skills, job_profile):

    candidate_skills = set(
        skill.lower() for skill in candidate_skills
    )

    required_skills = set(
        skill.lower()
        for skill in job_profile["required_skills"]
    )

    preferred_skills = set(
        skill.lower()
        for skill in job_profile["preferred_skills"]
    )

    matched_required = sorted(
        candidate_skills & required_skills
    )

    missing_required = sorted(
        required_skills - candidate_skills
    )

    matched_preferred = sorted(
        candidate_skills & preferred_skills
    )

    missing_preferred = sorted(
        preferred_skills - candidate_skills
    )

    required_score = (
        len(matched_required) /
        len(required_skills) * 100
        if required_skills else 0
    )

    preferred_score = (
        len(matched_preferred) /
        len(preferred_skills) * 100
        if preferred_skills else 0
    )

    return {
        "required_score": required_score,
        "preferred_score": preferred_score,
        "matched_required": matched_required,
        "missing_required": missing_required,
        "matched_preferred": matched_preferred,
        "missing_preferred": missing_preferred
    }
```

```python
import re
from datetime import datetime

def extract_experience(resume_text):
    text = resume_text.lower()

    # -----------------------------------------
    # 1. Explicit experience statements
    # -----------------------------------------
    explicit_patterns = [
        r'(\d+(?:\.\d+)?)\+?\s*(?:years?|yrs?)\s*(?:of)?\s*(?:professional\s+)?(?:work\s+)?experience',
        r'(\d+(?:\.\d+)?)\+?\s*(?:years?|yrs?)\s*(?:of)?\s*(?:relevant\s+)?experience',
    ]

    for pattern in explicit_patterns:
        matches = re.findall(pattern, text, flags=re.IGNORECASE)

        if matches:
            values = [float(x) for x in matches]
            return max(values)

    # -----------------------------------------
    # 2. Years + months experience
    # -----------------------------------------
    year_month_pattern = (
        r'(\d+(?:\.\d+)?)\s*years?'
        r'(?:\s*(?:and|,)?\s*(\d+)\s*months?)?'
    )

    matches = re.findall(year_month_pattern, text, flags=re.IGNORECASE)

    if matches:
        values = []

        for years, months in matches:
            years = float(years)
            months = float(months) if months else 0
            values.append(years + months / 12)

        if values:
            return max(values)

    # -----------------------------------------
    # 3. Date ranges: Month Year -> Current
    # -----------------------------------------
    month_pattern = (
        r'(jan(?:uary)?|feb(?:ruary)?|mar(?:ch)?|apr(?:il)?|'
        r'may|jun(?:e)?|jul(?:y)?|aug(?:ust)?|sep(?:tember)?|'
        r'sept(?:ember)?|oct(?:ober)?|nov(?:ember)?|dec(?:ember)?)'
        r'\s+(\d{4})'
        r'\s*(?:-|–|—|to)\s*'
        r'(current|present)'
    )

    matches = re.findall(month_pattern, text, flags=re.IGNORECASE)

    if matches:
        month_map = {
            'jan': 1, 'january': 1,
            'feb': 2, 'february': 2,
            'mar': 3, 'march': 3,
            'apr': 4, 'april': 4,
            'may': 5,
            'jun': 6, 'june': 6,
            'jul': 7, 'july': 7,
            'aug': 8, 'august': 8,
            'sep': 9, 'september': 9,
            'sept': 9, 'oct': 10, 'october': 10,
            'nov': 11, 'november': 11,
            'dec': 12, 'december': 12
        }

        experience_values = []

        for month, year, _ in matches:
            start_month = month_map[month.lower()]
            start_year = int(year)

            current_date = datetime.now()

            start_date = datetime(
                start_year,
                start_month,
                1
            )

            months = (
                (current_date.year - start_date.year) * 12
                + current_date.month - start_date.month
            )

            experience_values.append(months / 12)

        if experience_values:
            return round(max(experience_values), 2)

    # -----------------------------------------
    # 4. Year -> Current
    # Handles:
    # 2014 to Current
    # 2014 - Present
    # -----------------------------------------
    year_current_pattern = (
        r'(\d{4})\s*(?:-|–|—|to)\s*'
        r'(?:current|present)'
    )

    matches = re.findall(
        year_current_pattern,
        text,
        flags=re.IGNORECASE
    )

    if matches:
        current_year = datetime.now().year

        values = [
            current_year - int(year)
            for year in matches
        ]

        return round(max(values), 2)

    # -----------------------------------------
    # Nothing detected
    # -----------------------------------------
    return 0.0
```

```python
def calculate_experience_match(detected_experience, required_experience):
    if required_experience <= 0:
        return 100.0

    match = (detected_experience / required_experience) * 100

    return min(match, 100.0)
```

```python
import re

def extract_education(text):
    text = text.lower()

    # Normalize common formatting
    text = re.sub(r'[\n\r]+', ' ', text)
    text = re.sub(r'\s+', ' ', text)

    education = {
        "degrees": [],
        "fields": []
    }

    # -----------------------------
    # Degree detection
    # -----------------------------
    degree_patterns = {
        "b.tech": r'\bb\.?\s*tech\b|bachelor\s+of\s+technology',
        "b.e": r'\bb\.?\s*e\.?\b|bachelor\s+of\s+engineering',
        "b.sc": r'\bb\.?\s*sc\.?\b|bachelor\s+of\s+science',
        "bca": r'\bbca\b|bachelor\s+of\s+computer\s+applications',
        "m.tech": r'\bm\.?\s*tech\b|master\s+of\s+technology',
        "m.e": r'\bm\.?\s*e\.?\b|master\s+of\s+engineering',
        "m.sc": r'\bm\.?\s*sc\.?\b|master\s+of\s+science',
        "mba": r'\bmba\b|master\s+of\s+business\s+administration',
        "bachelor": r'\bbachelor(?:\'s)?\b',
        "master": r'\bmaster(?:\'s)?\b',
        "diploma": r'\bdiploma\b',
        "phd": r'\bph\.?\s*d\.?\b|doctorate'
    }

    for degree, pattern in degree_patterns.items():
        if re.search(pattern, text, re.IGNORECASE):
            education["degrees"].append(degree)

    # -----------------------------
    # Field detection
    # -----------------------------
    field_keywords = [
        "computer science",
        "computer engineering",
        "information technology",
        "information systems",
        "data science",
        "artificial intelligence",
        "machine learning",
        "software engineering",
        "electronics engineering",
        "electrical engineering",
        "mechanical engineering",
        "civil engineering",
        "business administration",
        "business management",
        "marketing",
        "finance",
        "accounting",
        "mathematics",
        "statistics",
        "physics",
        "chemistry"
    ]

    for field in field_keywords:
        if field in text:
            education["fields"].append(field)

    return education
```

```python
def calculate_education_match(candidate_education, required_education):
    candidate_fields = set(
        field.lower()
        for field in candidate_education["fields"]
    )

    required_fields = set(
        field.lower()
        for field in required_education
    )

    # No education requirement
    if not required_fields:
        return 100.0

    # Exact field match
    exact_matches = candidate_fields.intersection(required_fields)

    if exact_matches:
        return 100.0

    # Related fields
    related_fields = {
        "computer science": {
            "computer engineering",
            "software engineering",
            "information technology",
            "information systems",
            "artificial intelligence",
            "data science"
        },

        "information technology": {
            "computer science",
            "computer engineering",
            "information systems",
            "software engineering",
            "data science"
        },

        "data science": {
            "computer science",
            "statistics",
            "mathematics",
            "artificial intelligence",
            "machine learning",
            "information technology"
        }
    }

    for candidate_field in candidate_fields:
        for required_field in required_fields:

            if (
                candidate_field in related_fields
                and required_field in related_fields[candidate_field]
            ):
                return 70.0

    # Degree exists but field is unrelated
    if candidate_education["degrees"]:
        return 20.0

    # Nothing detected
    return 0.0
```

```python
import re

def extract_skills(text, skill_list):
    text = text.lower()

    found_skills = []

    for skill in skill_list:
        skill_lower = skill.lower()

        # Escape skill so special characters don't break regex
        pattern = r'(?<!\w)' + re.escape(skill_lower) + r'(?!\w)'

        if re.search(pattern, text):
            found_skills.append(skill)

    return found_skills
```

```python
def calculate_skills_match(
    matched_required,
    required_skills,
    matched_preferred,
    preferred_skills
):
    if not required_skills and not preferred_skills:
        return 100.0

    required_score = 0.0
    preferred_score = 0.0

    if required_skills:
        required_score = (
            len(matched_required) / len(required_skills)
        ) * 80

    if preferred_skills:
        preferred_score = (
            len(matched_preferred) / len(preferred_skills)
        ) * 20

    return round(required_score + preferred_score, 2)
```

```python
def extract_job_relevance(resume_text, job_profile):
    text = resume_text.lower()
    text = re.sub(r'\s+', ' ', text)

    job_title = job_profile["job_title"].lower().strip()
    domain = job_profile["domain"].lower().strip()

    score = 0
    matched_terms = []

    # --------------------------------
    # Exact job title
    # --------------------------------
    if re.search(
        r'(?<!\w)' + re.escape(job_title) + r'(?!\w)',
        text
    ):
        score += 50
        matched_terms.append(job_title)

    # --------------------------------
    # Exact domain
    # --------------------------------
    if re.search(
        r'(?<!\w)' + re.escape(domain) + r'(?!\w)',
        text
    ):
        score += 30
        matched_terms.append(domain)

    # --------------------------------
    # Related terms with weights
    # --------------------------------
    related_terms = {
        "data scientist": 40,
        "machine learning": 25,
        "artificial intelligence": 20,
        "predictive modeling": 20,
        "data analyst": 10,
        "data analysis": 10,
        "statistical modeling": 10,
        "data analytics": 10
    }

    related_score = 0

    for term, weight in related_terms.items():

        # Don't double-count exact title/domain
        if term == job_title or term == domain:
            continue

        if re.search(
            r'(?<!\w)' + re.escape(term) + r'(?!\w)',
            text
        ):
            matched_terms.append(term)
            related_score += weight

    score += min(related_score, 50)

    return min(score, 100.0), matched_terms
```

```python
def calculate_overall_score(
    experience_match,
    education_match,
    skills_match,
    relevance_score
):
    score = (
        experience_match * 0.15 +
        education_match * 0.20 +
        skills_match * 0.45 +
        relevance_score * 0.20
    )

    return round(score, 2)
```

```python
import re

def normalize_skill_text(text):
    text = str(text).lower()

    # Normalize common separators
    text = text.replace("-", " ")
    text = text.replace("_", " ")
    text = text.replace("/", " ")

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text


def detailed_skill_match(resume_text, required_skills, preferred_skills=None):

    if preferred_skills is None:
        preferred_skills = []

    resume_text = normalize_skill_text(resume_text)

    # Common skill aliases
    skill_aliases = {
        "python": ["python", "python programming"],
        "sql": ["sql", "structured query language"],
        "pandas": ["pandas"],
        "numpy": ["numpy"],
        "data analysis": [
            "data analysis",
            "data analytics",
            "data analyst",
            "data analysis using python"
        ],
        "statistics": [
            "statistics",
            "statistical analysis",
            "statistical methods"
        ],
        "data visualization": [
            "data visualization",
            "data visualisation",
            "data visualization using python",
            "data visualization tools"
        ],
        "machine learning": [
            "machine learning",
            "machine learning techniques",
            "ml techniques"
        ],
        "predictive modeling": [
            "predictive modeling",
            "predictive modelling",
            "predictive analytics"
        ],
        "scikit-learn": [
            "scikit learn",
            "scikit-learn",
            "sklearn"
        ]
    }

    def skill_found(skill):

        skill_normalized = normalize_skill_text(skill)

        possible_forms = skill_aliases.get(
            skill_normalized,
            [skill_normalized]
        )

        for form in possible_forms:
            form = normalize_skill_text(form)

            # Word-boundary matching
            pattern = r"\b" + re.escape(form) + r"\b"

            if re.search(pattern, resume_text):
                return True

        return False

    # -----------------------------------------
    # Required skills
    # -----------------------------------------

    matched_required = []
    missing_required = []

    for skill in required_skills:

        if skill_found(skill):
            matched_required.append(skill)
        else:
            missing_required.append(skill)

    # -----------------------------------------
    # Preferred skills
    # -----------------------------------------

    matched_preferred = []

    for skill in preferred_skills:

        if skill_found(skill):
            matched_preferred.append(skill)

    # -----------------------------------------
    # Scores
    # -----------------------------------------

    required_score = (
        len(matched_required) / len(required_skills) * 100
        if required_skills
        else 100
    )

    preferred_score = (
        len(matched_preferred) / len(preferred_skills) * 100
        if preferred_skills
        else 0
    )

    return {
        "matched_required": matched_required,
        "missing_required": missing_required,
        "matched_preferred": matched_preferred,
        "required_score": round(required_score, 2),
        "preferred_score": round(preferred_score, 2)
    }
```

```python
def calculate_final_skill_score(skill_details):
    required_score = skill_details["required_score"]
    preferred_score = skill_details["preferred_score"]

    # If there are no preferred skills, use only required skills
    if len(skill_details["matched_preferred"]) == 0:
        return required_score

    # Otherwise, required skills have higher weight
    final_score = (required_score * 0.8) + (preferred_score * 0.2)

    return round(final_score, 2)
```

```python
def calculate_overall_match(
    experience_match,
    education_match,
    skills_match,
    job_relevance
):
    overall = (
        experience_match * 0.25 +
        education_match * 0.20 +
        skills_match * 0.40 +
        job_relevance * 0.15
    )

    return round(overall, 2)
```

```python
def screen_candidate(
    resume_text,
    required_skills,
    preferred_skills,
    required_experience,
    required_education,
    job_profile
):
    # -----------------------------
    # Experience
    # -----------------------------
    experience = extract_experience(resume_text)

    experience_match = min(
        (experience / required_experience) * 100,
        100
    ) if required_experience > 0 else 100

    # -----------------------------
    # Education
    # -----------------------------
    education = extract_education(resume_text)

    education_match = calculate_education_match(
        education,
        required_education
    )

    # -----------------------------
    # Skills
    # -----------------------------
    skill_details = detailed_skill_match(
        resume_text,
        required_skills,
        preferred_skills
    )

    final_skill_score = calculate_final_skill_score(
        skill_details
    )

    # -----------------------------
    # Job Relevance
    # -----------------------------
    relevance_score, relevant_terms = extract_job_relevance(
        resume_text,
        job_profile
    )

    # -----------------------------
    # Overall Score
    # -----------------------------
    final_overall_match = round(
        experience_match * 0.25 +
        education_match * 0.20 +
        final_skill_score * 0.40 +
        relevance_score * 0.15,
        2
    )

    return {
        "experience": experience,
        "experience_match": round(experience_match, 2),

        "education": education,
        "education_match": round(education_match, 2),

        "skill_details": skill_details,
        "skills_match": final_skill_score,

        "relevant_terms": relevant_terms,
        "job_relevance": round(relevance_score, 2),

        "overall_match": final_overall_match
    }
```

```python
def generate_explanation(result):
    
    matched_skills = result["skill_details"]["matched_required"]
    missing_skills = result["skill_details"]["missing_required"]
    relevant_terms = result["relevant_terms"]

    explanation = []

    # Experience
    if result["experience_match"] >= 100:
        explanation.append(
            f"Meets the required experience with "
            f"{result['experience']} years of experience."
        )
    else:
        explanation.append(
            f"Has {result['experience']} years of experience "
            f"against the required experience."
        )

    # Education
    if result["education_match"] >= 100:
        explanation.append(
            "Education is highly aligned with the job requirements."
        )
    elif result["education_match"] >= 70:
        explanation.append(
            "Education is related to the job requirements."
        )
    else:
        explanation.append(
            "Education has limited alignment with the job requirements."
        )

    # Skills
    if matched_skills:
        explanation.append(
            f"Matches {len(matched_skills)} required skills."
        )

    if missing_skills:
        explanation.append(
            "Missing key skills: " +
            ", ".join(missing_skills) + "."
        )

    # Relevance
    if relevant_terms:
        explanation.append(
            "Relevant terms found: " +
            ", ".join(relevant_terms) + "."
        )

    return " ".join(explanation)
```
